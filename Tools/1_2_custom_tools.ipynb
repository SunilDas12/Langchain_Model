Method-1: Using @tool decorator

```python
from langchain_community.tools import tool
```

```python
# step-1 - create a function
def multiply(a, b):
    """Multiply two numbers"""
    return a*b
```

```python
# step-2 - add type hints
def multiply(a: int, b: int):
    """Multiply two numbers"""
    return a*b
```

```python
# step-3 - add tool decorator
@tool
def multiply(a: int, b: int)-> int:
    """Multiply two numbers"""
    return a*b
```

```python
result=multiply.invoke({"a":3, "b":2})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
#LLM will see this
print(multiply.args_schema.model_json_schema())
```

Method-2: Using StructuredTool

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int=Field(required=True, description="The first number to add")
    b: int=Field(required=True, description="The second number to add")
```

```python
def multiply_func(a: int, b:int)->int:
    return a * b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="multiply two number",
    args_schema=MultiplyInput
)
```

```python
result=multiply_tool.invoke({"a":3, "b":2})
print(result)
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

Method 3- Using BaseTool Class

```python
from langchain_community.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int=Field(required=True, description="The first number to add")
    b: int=Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str="multiply"
    description: str="multiply two numbers"
    args_schema: Type[BaseModel]=MultiplyInput

    def _run(self, a:int , b:int) ->int:
        return a*b
```

```python
multiply_tool_base=MultiplyTool()
```

```python
result=multiply_tool_base.invoke({"a":3, "b":2})
```

```python
print(result)
print(multiply_tool_base.name)
print(multiply_tool_base.description)
print(multiply_tool_base.args)
```
