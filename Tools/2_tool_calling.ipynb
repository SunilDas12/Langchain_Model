All models don't have have the capabilty to call the tools. There are some specific model which can call the tools

```python
import os
os.environ['OPENAI_API_KEY']=''
os.environ['OPENAI_API_VERSION']=''
os.environ['OPENAI_AZURE_ENDPOINT']=''
os.environ['OPENAI_AZURE_MODEL']=''
```

```python
from langchain_openai import AzureChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

1-Tool Create

```python
@tool
def multiply(a: int, b: int)->int:
    """Given 2 numbers a and b this tool return their product"""
    return a * b
```

```python
print(multiply.invoke({'a':3, 'b':4}))
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

2-Tool binding

```python
llm=AzureChatOpenAI(
        api_key=os.environ['OPENAI_API_KEY'],
        api_version=os.environ['OPENAI_API_VERSION'],
        azure_endpoint=os.environ['OPENAI_AZURE_ENDPOINT'],
        model_name=os.environ['OPENAI_AZURE_MODEL'],
        temperature=0.4,
        max_tokens=1000,
        seed=42,
)
```

```python
llm_with_tools=llm.bind_tools([multiply])
```

3-Tool Calling

```python
llm_with_tools.invoke('hi how are you') # here it didn't find anything to call the respective tool
```

```python
query=HumanMessage('can you multiply 3 with 100')
messages=[query]
messages
```

```python
llm_result=llm_with_tools.invoke(messages)
llm_result
```

```python
messages.append(llm_result)
messages
```

4-Tool Execution


```python
llm_result.tool_calls[0]   # For debugging
```

```python
tool_result=multiply.invoke(llm_result.tool_calls[0])
tool_result
```

```python
messages.append(tool_result)
messages
```

```python
llm_with_tools.invoke(messages).content
```
